```python
list_1=[124,344,5,75,458,4578,4578,12,89]
xx_1=[]
xx=str(list_1).split()
print("print xx", xx)
#print(len(xx[1]))
sum1=0
j=0
xxx=[]
for j in range(len(list_1)):
    for i in range(len(xx[j])-1):
        if j==0 and i>=1:
            sum1=sum1+int(xx[j][i])
        elif j > 0:
            sum1 = sum1 +int(xx[j][i])
    xxx.append(sum1)
    j=j+1
print(xxx)  

yy=[]
import numpy as np
for i in range(len(xxx)):
    if i==0:
        y=xxx[0]
        yy.append(y)
    elif i>=1:
        y=(xxx[i]-xxx[i-1])
        i=i+1
        yy.append(y)   
print("print yy", yy)

prime_1=[]
not_prime=[]
for jj in range(len(yy)):
    if yy[jj]>1:
        for i in range(2,yy[jj]):
            if (yy[jj]%i)==0:
                not_prime.append(yy[jj])
                break
        else:
            prime_1.append(yy[jj])
print("List of not prime numbers", not_prime)
print("list of prime numbers", prime_1)
```
