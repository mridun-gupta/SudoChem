```python
# Question: Given list_1 = [457, 789, 4587, 451254, 452452145231, 45, 458796545]
# Task: Add digits of each number and check if the sum is prime or not
# - Make a list of sum of digits
# - Make a list of prime sums
# - Make a list of not prime sums

list_1 = [457, 789, 4587, 451254, 452452145231, 45, 458796545]

# Step 1: Calculate sum of digits for each number
sum_of_digits = []
for num in list_1:
	digit_sum = sum(int(digit) for digit in str(num))
	sum_of_digits.append(digit_sum)
print("Sum of digits:", sum_of_digits)

# Step 2: Check which sums are prime
prime_sum = []
not_prime_sum = []
for s in sum_of_digits:
	if s > 1:
		is_prime = True
		for i in range(2, s):
			if s % i == 0:
				is_prime = False
				break
		if is_prime:
			prime_sum.append(s)
		else:
			not_prime_sum.append(s)
	else:
		not_prime_sum.append(s)

print("List of prime sums:", prime_sum)
print("List of not prime sums:", not_prime_sum)
```
