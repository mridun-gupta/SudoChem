```python
import pandas as pd
import seaborn as sns
%matplotlib inline
tips = pd.read_csv('tip.csv')
tips
```

```python
#distplot
```

```python
sns.displot(tips['total_bill'])
```

```python
sns.displot(tips['total_bill'], kde=False, bins=30)
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='scatter')
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='hex')
```

```python
sns.jointplot(x='total_bill', y='tip', data=tips, kind='reg')
```

```python
sns.pairplot(tips)
```

```python
sns.pairplot(tips, hue='sex', palette= 'coolwarm')
```

```python
# categorical plots
```

```python
sns.barplot(x='sex', y='total_bill', data = tips)
```

```python
import numpy as np
```

```python
sns.barplot(x='sex', y='total_bill', data = tips, estimator= np.std)
```

```python
sns.countplot(x='sex', data= tips)
```

```python
sns.boxplot(x='day', y='total_bill', data=tips, palette='rainbow')
```

```python
sns.boxplot(x='day', y='total_bill', hue= 'smoker', data=tips, palette='rainbow')
```

```python
sns.stripplot(x='day', y='total_bill', data=tips)
```

```python
sns.stripplot(x='day', y='total_bill', hue='sex' ,data=tips)
```

```python
sns.swarmplot(x='day', y='total_bill', data=tips)
```

```python
sns.swarmplot(x='day', y='total_bill', hue= 'sex', data=tips)
```

```python
sns.violinplot(x='day',y='total_bill', data=tips, palette = 'rainbow')
sns.swarmplot(x='day',y='total_bill', data=tips, color='black', size=3)
```
