```python
import numpy as np
import pandas as pd
import scipy as sp
import matplotlib.pyplot as plt
%matplotlib inline
# plt.show()
```

```python
x = np.linspace(0,5,11)
y = x**2
x
```

```python
y
```

```python
plt.plot(x,y,'r') # r is for red color
plt.xlabel('X axis title')
plt.ylabel('Y axis title')
plt.title('Title for graph')
# plt.show()
```

```python
plt.scatter(x,y) # r is for red color
plt.xlabel('X axis title')
plt.ylabel('Y axis title')
plt.title('Title for graph')
# plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(x,y)
plt.xlabel('X axis title')
plt.ylabel('Y axis title')
plt.title('Title for graph1')
plt.subplot(1,2,2)
plt.plot(y,x)
plt.xlabel('X axis title')
plt.ylabel('Y axis title')
plt.title('Title for graph2')
```

```python
plt.subplot(1,2,1)
plt.scatter(x,y)
plt.subplot(1,2,2)
plt.scatter(y,x)
```

```python
# object oriented method
```

```python
# create figure (empty canvas)
fig = plt.figure()

# add set of axis to the figure
axes = fig.add_axes([0.1, 0.1, 0.8, 0.8]) # left , right, width, height (range 0 and 1)

axes.plot(x,y,'b')
axes.set_xlabel('X axis title')
axes.set_ylabel('Y axis title')
axes.set_title('Title for the graph')
```

```python
fig = plt.figure()

axes1 = fig.add_axes([0.1,0.1,0.8,0.8])
axes2 = fig.add_axes([0.2,0.5,0.4,0.3])

# larger graph
axes1.plot(x,y)
axes1.set_xlabel('X axis title 1')
axes1.set_ylabel('Y axis title 1')
axes1.set_title('Title for the graph 1')

#smaller graph
axes2.plot(y,x)
axes2.set_xlabel('X axis title 2')
axes2.set_ylabel('Y axis title 2')
axes2.set_title('Title for the graph 2')
```

```python
fig, axes = plt.subplots(nrows=1, ncols=3)
```

```python
for ax in axes:
    ax.plot(x,y)
    ax.set_xlabel('x')
    ax.set_ylabel('y')
    ax.set_title('title')

fig.tight_layout()
fig
```

```python
fig = plt.figure(figsize=(8,4), dpi = 100)
```

```python
fig, axes = plt.subplots(figsize=(8,8))

axes.plot(x,y,'r')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('title')
```

```python
fig.savefig('demofig.png', dpi=300)
```

```python
fig, axes = plt.subplots(figsize=(8,8))

axes.scatter(x,x**2, label='x**2')
axes.scatter(x,x**3, label='x**3')
axes.set_xlabel('x')
axes.set_ylabel('y')
axes.set_title('title')
axes.legend()
```
