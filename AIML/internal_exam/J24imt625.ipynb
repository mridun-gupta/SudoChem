```python
X=df2[['LogP1', 'MR1', 'HL1', 'Tpsa1', 'CLog P1', 'CMR1','LogS1']]
y=df2['MIC90 (micro M)']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.4, random_state = 101)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lm = LinearRegression()
```

```python
lm.fit(X_train,y_train)
```

```python
print(lm.intercept_)
```

```python
y_pred = lm.predict(X_test)
```

```python
plt.scatter(y_test, y_pred)
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
dt = DecisionTreeRegressor()
```

```python
dt.fit(X_train,y_train)
```

```python
y1_pred = dt.predict(X_test)
```

```python
plt.scatter(y_test, y1_pred)
```

```python
from sklearn import metrics
from sklearn.metrics import r2_score
```

```python
print('MAE', metrics.mean_absolute_error(y_test,y1_pred))
print('MSE', metrics.mean_squared_error(y_test,y1_pred))
print('RMSE', np.sqrt(metrics.mean_squared_error(y_test,y1_pred)))
print('R2', r2_score(y_test,y1_pred))
```
